```python
# Function to extract only the first numeric part
import re

def extract_numbers(value):
    if pd.isna(value):
        return None
    match = re.match(r"(\d{1,2})", str(value))
    return int(match.group(1)) if match else None
```

```python
# Split Arrival_Time into hour and minute we can also use lambda function
# First split into hour and minute from the original string
final_df["Arrival_hour"] = final_df["Arrival_Time"].str.split(":").str[0].apply(extract_numbers)
final_df["Arrival_minute"] = final_df["Arrival_Time"].str.split(":").str[1].apply(extract_numbers)
```

```python
final_df["Arrival_Time"] = final_df["Arrival_hour"]  # or any cleaned format you want
```

```python
import re

def duration_to_minutes(value):
    if pd.isna(value):
        return None
    hours = 0
    minutes = 0
    match_hours = re.search(r'(\d+)h', value)
    match_minutes = re.search(r'(\d+)m', value)
    if match_hours:
        hours = int(match_hours.group(1))
    if match_minutes:
        minutes = int(match_minutes.group(1))
    return hours * 60 + minutes

final_df["Duration_minutes"] = final_df["Duration"].apply(duration_to_minutes)

```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

ohe = OneHotEncoder(sparse_output=False)  # Updated parameter

# Columns to encode
cols_to_encode = ['Airline', 'Source', 'Destination', 'Additional_Info']

encoded_array = ohe.fit_transform(final_df[cols_to_encode])
encoded_df = pd.DataFrame(encoded_array, columns=ohe.get_feature_names_out(cols_to_encode), index=final_df.index)

# Drop original and join new encoded columns
final_df = pd.concat([final_df.drop(cols_to_encode, axis=1), encoded_df], axis=1)
```

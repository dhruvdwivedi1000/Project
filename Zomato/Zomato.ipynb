```python
df_mer.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size() #by size() we can see alll the value ,Purpose: Counts the number of rows in each group.
#'A groupby operation involves some combination of splitting the
#object, applying a function, and combining the results. This can be
#used to group large amounts of data and compute operations on these
#groups.
```

```python
rating = df_mer.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns = {0:'Rating Count'})
#Purpose: Converts the group index back into regular columns of a new DataFrame.
#Without this, the grouped columns remain as a MultiIndex in the Series.
#.rename(columns={0: 'Rating Count'})
#Purpose: The .size() method names the count column as 0 by default.
#This step renames that column to something more meaningful: 'Rating Count'.
```

```python
rating
```

## Obervation
1. When rating is between 4.5 to .4.9 ----> Excellent
2. when rating are between 4.0 to 3.9 ----> Very good
3. when rating are between 3.5 to 3.9 ---->  good
4. when rating are between 3.0 to 3.4 ----> Average
5. when rating are between 2.5 to 2.9 ----> Average
6. when rating are between 2.0 to 2.4 ----> Poor

```python
rating.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize'] = (12, 6) #it adjust the figure of the graph (h,b)
sns.barplot(x = 'Aggregate rating' , y = 'Rating Count',data= rating)
plt.show()
```

```python
sns.barplot(x = 'Aggregate rating' , y = 'Rating Count',hue = 'Rating color',data= rating ,palette =[ 'white','red' , 'yellow','green','green' ])
plt.show()
```

### Observation
  1. Not rated count is very high
  2. maximum no. of rating are between 2.5 to 3.4

```python
#count plot
sns.countplot(x='Rating color',data = rating,palette =[ 'black','red' , 'yellow','green','lightgreen','lightblue' ])
plt.show()
```
